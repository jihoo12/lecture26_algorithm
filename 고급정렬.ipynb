# 10814

```python
import sys


input = sys.stdin.read

def solve():
    lines = input().splitlines()
    n = int(lines[0])
    members = []
    for i in range(1, n + 1):
        age, name = lines[i].split()
        members.append((int(age), name))
    
    members.sort(key=lambda x: x[0])
    
    for age, name in members:
        print(f"{age} {name}")

if __name__ == "__main__":
    solve() 
```

# 11650

```python
import sys
input = sys.stdin.read

def solve():
    data = input().split()
    n = int(data[0])
    points = []
    idx = 1
    for _ in range(n):
        x = int(data[idx])
        y = int(data[idx+1])
        points.append((x, y))
        idx += 2
    points.sort(key=lambda p: (p[0], p[1]))
    for x, y in points:
        print(f"{x} {y}")

if __name__ == "__main__":
    solve()
```

# 1181

```python
import sys
input = sys.stdin.read

def solve():
    lines = input().splitlines()
    if not lines:
        return
    
    n = int(lines[0])
    words = list(set(lines[1:n+1]))
    words.sort(key=lambda x: (len(x), x))
    for word in words:
        print(word)

if __name__ == "__main__":
    solve()
```

# 10825

```python
import sys
input = sys.stdin.read

def solve():
    lines = input().splitlines()
    n = int(lines[0])
    
    students = []
    for i in range(1, n + 1):
        name, kor, eng, math = lines[i].split()
        students.append((name, int(kor), int(eng), int(math)))
        
    students.sort(key=lambda x: (-x[1], x[2], -x[3], x[0]))
    
    for student in students:
        print(student[0])

if __name__ == "__main__":
    solve()
```
